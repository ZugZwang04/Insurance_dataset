```python
import pandas as pd

df = pd.read_csv('../insurance_dataset_400_raw.csv')
```

```python
df.info()
```

```python
df.sample(5)
```

```python
df[['cust_first_name', 'cust_last_name']] = df['customer_name'].str.lower().str.strip().str.title().str.split(" ", n = 1, expand = True)
```

```python
df[['agent_first_name', 'agent_last_name']] = df['agent_name'].str.lower().str.strip().str.title().str.split(" ", n = 1, expand = True)
```

```python
cols = ['province', 'payment_schedule', 'policy_status', 'payment_status']

df[cols] = df[cols].apply(lambda x: x.astype(str).str.lower().str.strip().str.title())
```

```python
cust_id = 1
for cust in df['customer_name'].unique():
    df.loc[df['customer_name'] == cust, 'customer_id'] = cust_id
    cust_id += 1
```

```python
ins_id = 1
for insurance in df['insurance_service'].unique():
    df.loc[df['insurance_service'] == insurance, 'insurance_id'] = ins_id
    ins_id += 1
```

```python
agent_id = 1
for agent in df['agent_name'].unique():
    df.loc[df['agent_name'] == agent, 'agent_id'] = agent_id
    agent_id += 1
```

```python
province_id = 1
for province in df['province'].unique():
    df.loc[df['province'] == province, 'province_id'] = province_id
    province_id += 1
```

```python
df.columns
```

```python
df = df[[
    # not be included
    'customer_name', 'agent_name', 'premium_per_payment', 'agent_commission', 
    


    # policy table
    'policy_id', 'payment_schedule', 'policy_status', 'payment_status','start_date', 
    'claim_date', 'claim_amount', 'annual_premium', 'payments_per_year', 'total_premium',   # FK: 'insurance_id', 'agent_id', 'customer_id'
     
    # insurance table
    'insurance_id', 'insurance_service', 'policy_term_years',
    
    # agents table
    'agent_id', 'agent_first_name', 'agent_last_name',

    # customers table
    'customer_id', 'cust_first_name', 'cust_last_name', #FK: 'province_id'
    
    # provinces table
    'province_id', 'province', 'region'
]]
```

```python
for ins in df['insurance_service'].unique():
    cond = df['insurance_service'] == ins
    print(df.loc[cond, ['insurance_service', 'policy_term_years']])
```

```python
cols = ['insurance_id', 'agent_id', 'customer_id', 'province_id']
df[cols] = df[cols].apply(lambda col: col.astype('Int64'))
```

```python
df['start_date'] = pd.to_datetime(df['start_date'], dayfirst=True)  #     convert to datetime
df['start_date'] = df['start_date'].dt.strftime('%Y-%m-%d')         #     format as ISO 8601 (YYYY-MM-DD)
```

```python
df['claim_date'] = pd.to_datetime(df['claim_date'], dayfirst=True)  
df['claim_date'] = df['claim_date'].dt.strftime('%Y-%m-%d')         
```

```python
df.sample(5)
```

```python
provinces = df[['province_id', 'province', 'region']].drop_duplicates()
provinces.to_csv('../Normalize/provinces.csv', index=False)
```

```python
customers = df[['customer_id', 'province_id', 'cust_first_name', 'cust_last_name']].drop_duplicates()
customers.to_csv('../Normalize/customers.csv', index=False)
```

```python
services = df[['insurance_id', 'insurance_service', 'policy_term_years']].drop_duplicates()
services.to_csv('../Normalize/services.csv', index=False)
```

```python
agents = df[['agent_id', 'agent_first_name', 'agent_last_name']].drop_duplicates()
agents.to_csv('../Normalize/agents.csv')
```

```python
policies = df[['policy_id', 'customer_id', 'agent_id', 'insurance_id', 'payment_schedule', 'policy_status', 'payment_status','start_date', 'claim_date', 'claim_amount', 'annual_premium', 'payments_per_year', 'total_premium']].drop_duplicates()
policies.to_csv('../Normalize/policies.csv', index=False)
```

```python
from sqlalchemy import create_engine

# ----------------------------
# 1. Connect to MySQL
# ----------------------------
engine = create_engine(
    "mysql+pymysql://root:bettercallsaul@localhost:3307/db_insurance"
)

# ----------------------------
# 2. Clean Data
# ----------------------------
provinces = provinces.drop_duplicates(subset=["province_id"])
customers = customers.drop_duplicates(subset=["customer_id"])
agents = agents.drop_duplicates(subset=["agent_id"])
services = services.drop_duplicates(subset=["insurance_id"])
policies = policies.drop_duplicates(subset=["policy_id"])

# ----------------------------
# 3. Insert Into MySQL Tables
# ----------------------------
try:
    # Insert parent tables first
    provinces.to_sql("provinces", engine, if_exists="append", index=False)
    customers.to_sql("customers", engine, if_exists="append", index=False)
    agents.to_sql("agents", engine, if_exists="append", index=False)
    services.to_sql("services", engine, if_exists="append", index=False)

    # Insert child table last
    policies.to_sql("policies", engine, if_exists="append", index=False)

    print("✅ All data successfully loaded into MySQL!")

except Exception as e:
    print("❌ Error loading data:", e)
```
